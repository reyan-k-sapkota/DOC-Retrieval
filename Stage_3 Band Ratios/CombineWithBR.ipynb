```python
#TO Create final trainable CSV with respective band ratios
#For Chlorophyll only

import pandas as pd

chlorophyll_withoutBR= pd.read_csv("D:\Shukra_sir\MLAllTraining\ForTraining\Chlorophyll.csv") 
chlorophyll_withBR1 = pd.read_csv("D:\\Shukra_sir\\MLAllTraining\\ForTraining\\Band Ratio and Correlations\\For Chlorophyll\\NewChlorophyll_BandRatio1.csv")
chlorophyll_withBR2 = pd.read_csv("D:\\Shukra_sir\\MLAllTraining\\ForTraining\\Band Ratio and Correlations\\For Chlorophyll\\NewChlorophyll_BandRatio2.csv")
chlorophyll_withBR3 = pd.read_csv("D:\\Shukra_sir\\MLAllTraining\\ForTraining\\Band Ratio and Correlations\\For Chlorophyll\\NewChlorophyll_BandRatio3.csv")

doc_withoutBR = pd.read_csv("D:\Shukra_sir\MLAllTraining\ForTraining\DOC.csv")
doc_withBR1 = pd.read_csv("D:\\Shukra_sir\\MLAllTraining\\ForTraining\\Band Ratio and Correlations\\For DOC\\NewDOC_BandRatio1.csv")
doc_withBR2 = pd.read_csv("D:\\Shukra_sir\\MLAllTraining\\ForTraining\\Band Ratio and Correlations\\For DOC\\NewDOC_BandRatio2.csv")
doc_withBR3 = pd.read_csv("D:\\Shukra_sir\\MLAllTraining\\ForTraining\\Band Ratio and Correlations\\For DOC\\NewDOC_BandRatio3.csv")
```

```python
#Band Ratio with Highest Correlations, looked manually

"""
A. FOR CHLOROPHYLL
a. for BandRatio1: 
Top three are: 
i. "B2_minus_B5", 
ii. "B3_minus_B2", 
iii. "B2_minus_B4",

    
b. for Band Ratio2:
Top three are: 
i. "B3_div_B5", 
ii. "B2_div_B5", 
iii. "B2_div_B4"

    
c. for Band Ratio3:
Top three are: 
i. "ND_B5_B3",
ii. "ND_B2_B5",
iii."ND_B4_B2"

"""


"""
A. FOR DOC
a. for BandRatio1: 
Top three are: 
i. 
ii. 
iii.
b. for Band Ratio2:
Top three are: 
i. 
ii. 
iii.

c. for Band Ratio3:
Top three are: 
i. 
ii. 
iii.
"""
```

```python
"For Chlorophyll only"

top_BR1 = [
    "B2_minus_B5",
    "B3_minus_B2",
    "B2_minus_B4"
]

top_BR2 = [
    "B3_div_B5",
    "B2_div_B5",
    "B2_div_B4"
]

top_BR3 = [
    "ND_B5_B3",
    "ND_B2_B5",
    "ND_B4_B2"
]
```

```python
df_base = chlorophyll_withoutBR
df_BR1 = chlorophyll_withBR1
df_BR2 = chlorophyll_withBR2
df_BR3 = chlorophyll_withBR3

df_BR1_sel = df_BR1[top_BR1]
df_BR2_sel = df_BR2[top_BR2]
df_BR3_sel = df_BR3[top_BR3]
```

```python
df_BR1_sel
```

```python
df_final = pd.concat(
    [df_base, df_BR1_sel, df_BR2_sel, df_BR3_sel],
    axis=1
)

df_final
```

```python
output_file = "D:\\Shukra_sir\\MLAllTraining\\ForTraining\\TrainableWithBandRatios\\Chlorophyll_BR.csv" 
df_final.to_csv(output_file, index=False)
```
