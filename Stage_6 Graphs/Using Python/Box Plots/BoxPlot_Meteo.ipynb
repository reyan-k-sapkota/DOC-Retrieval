```python
df = pd.read_csv("D:\Shukra_sir\WQI_Research\Scripts and Data\StatisticsAnalysis\Mean\DOC_Meteo_DescStats.csv")
df.shape
```

```python
df.set_index(df.columns[0], inplace=True)

# Clean index text
df.index = df.index.str.strip()

print(df.index)
```

```python
cols = ['Temperature', 'DewPoint']
print(cols)
```

```python
labels = ['2m Temperature', '2m Dewpoint Temperature']

box_data = []
i = 0

for meteo in cols:
    stats = {
        'label': labels[i],
        'whislo': df.loc['min', meteo],
        'q1': df.loc['Q1_25%', meteo],
        'med': df.loc['Median_Q2_50%', meteo],
        'q3': df.loc['Q3_75%', meteo],
        'whishi': df.loc['max', meteo],
        'fliers': []
    }
    box_data.append(stats)
    i +=1


# -------------------------
# Create Figure
# -------------------------
fig, ax = plt.subplots(figsize=(12,6))

box = ax.bxp(
    box_data,
    showfliers=False,
    patch_artist=True,
    boxprops=dict(facecolor="#6495ED", alpha = 0.81 ,edgecolor="black", linewidth=0.93),
    medianprops=dict(color="#8B0000", linewidth=1.5),
    whiskerprops=dict(color="black", linewidth=1.2),
    capprops=dict(color="black", linewidth=1.2)
)

# -------------------------
# Axis Styling
# -------------------------
ax.set_xlabel("Meteorological Variables", fontsize=15, fontweight='bold')
ax.set_ylabel("Kelvin (K)", fontsize=15, fontweight='bold')
#ax.set_title("Distribution of Spectral Bands (B1–B12)", fontsize=18, fontweight='bold')

ax.tick_params(axis='both', labelsize=12)

# Light horizontal grid only
ax.yaxis.grid(True, linestyle='--', alpha=0.6)
ax.xaxis.grid(False)

# Thicker border
for spine in ax.spines.values():
    spine.set_linewidth(1.2)

plt.tight_layout()
plt.show()
```
