```python
import matplotlib.pyplot as plt
```

```python
box_data = {
    #'label': 'DOC (mg/lit)',
    'whislo': 0.5,
    'q1': 2,
    'med': 2.9,
    'q3': 3.8,
    'whishi': 15.2,
    'fliers': []
}


# -------------------------
# Create Figure
# -------------------------
fig, ax = plt.subplots(figsize=(9,6))

box = ax.bxp(
    [box_data],
    showfliers=False,
    patch_artist=True,
    boxprops=dict(facecolor="#6495ED", alpha = 0.81 ,edgecolor="black", linewidth=0.93),
    medianprops=dict(color="#8B0000", linewidth=1.5),
    whiskerprops=dict(color="black", linewidth=1.2),
    capprops=dict(color="black", linewidth=1.2)
)

# -------------------------
# Axis Styling
# -------------------------
ax.set_xlabel("Dissolved Organic Carbon (DOC)", fontsize=15, fontweight='bold')
ax.set_ylabel("Concentration of Dissolved Organic Carbon (mg/L as C)", fontsize=12, fontweight='bold')
#ax.set_title("Distribution of Spectral Bands (B1–B12)", fontsize=18, fontweight='bold')

ax.tick_params(axis='both', labelsize=12)

# Light horizontal grid only
ax.yaxis.grid(True, linestyle='--', alpha=0.6)
ax.xaxis.grid(False)

# Thicker border
for spine in ax.spines.values():
    spine.set_linewidth(1.2)

plt.tight_layout()
plt.show()
```
