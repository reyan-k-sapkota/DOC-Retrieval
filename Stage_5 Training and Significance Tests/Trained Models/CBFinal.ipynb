```python
#FOR AUGMENTED_NEW

import pandas as pd
import numpy as np

aug_factor = 3 #augmented has been done initially for aug_factor = 2, then for aug_factor = 3

doc_train_augmented = pd.read_csv (f"D:\Shukra_sir\MLAllTraining\ForTraining\TrainableWithBandRatios\AugmentedWithBR\DOC_AUG_WITH_BR\DOCWithBR_TRAIN_AUG{aug_factor}.csv")
doc_test = pd.read_csv(f"D:\Shukra_sir\MLAllTraining\ForTraining\TrainableWithBandRatios\AugmentedWithBR\DOC_AUG_WITH_BR\DOCWithBR_TEST_AUG{aug_factor}.csv")

doc_train_augmented.shape, doc_test.shape
```

```python
doc_train_augmented ['result_DOC_log'] = np.log1p (doc_train_augmented ['result_DOC'])
doc_test ['result_DOC_log'] = np.log1p (doc_test ['result_DOC'])
doc_train_augmented
```

```python
TARGET = "result_DOC_log" #FOR DOC
TARGET1 = "result_DOC"

input_features = [#'B1',
                  'B2', 'B3', 'B4', 'B5', 'B6', 'B7', 'B8', 'B8A', 
                'B9',
               "B4_minus_B3",
               
               "B4_div_B3", 
               
               "ND_B4_B3", 
               
               'Temperature', 'DewPoint', 'v10n', 
               'doy_sin', 'doy_cos', 
                
               'latitude', 'longitude'
               ]

X_train = doc_train_augmented[input_features]
y_train = doc_train_augmented[TARGET]

X_test = doc_test[input_features]
y_test = doc_test[TARGET]

X_train.shape, y_train.shape, X_test.shape, y_test.shape

```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error, root_mean_squared_error
import catboost
print(catboost.__version__)
```

```python
from catboost import CatBoostRegressor
import numpy as np

X = np.random.rand(1000, 10)
y = np.random.rand(1000)

model = CatBoostRegressor(
    iterations=50,
    task_type="GPU",
    devices='0',
    verbose=0
)

model.fit(X, y)
print("CatBoost GPU training successful ✅")
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled= scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python

baseline_CBR = CatBoostRegressor(
    iterations=3600,         # baseline, can increase later
    learning_rate=0.03,      # baseline
    depth=4,                 # baseline
    l2_leaf_reg=3,            # L2 regularization, prevents overfitting
    task_type="CPU",         # GPU training
    devices='0',             # first GPU
    random_seed=42,
    verbose=100              # log progress every 100 iterations
)

baseline_CBR.fit(X_train_scaled, y_train)

y_pred = baseline_CBR.predict(X_test_scaled)

print("R² for DOC for log scaled | CatBoost Regressor:", r2_score(y_test, y_pred))

y_pred_orig = np.expm1(y_pred)
y_test_real = np.expm1(y_test)
print("R² for DOC on normal scaled | CatBoost Regressor:", r2_score(y_test_real, y_pred_orig))


# Back-transform predictions and true values
y_pred_real = np.expm1(y_pred)
y_test_real = np.expm1(y_test)


print("R² for DOC on normal:", r2_score(y_test_real, y_pred_real))

# ---- Metrics on normal scale ----
mse  = mean_squared_error(y_test_real, y_pred_real)
rmse = np.sqrt(mse)
mae  = mean_absolute_error(y_test_real, y_pred_real)
r2   = r2_score(y_test_real, y_pred_real)

print("Final metrics on NORMAL scale (after inverse log):")
print(f"R²   : {r2:.4f}")
print(f"MSE  : {mse:.4f}")
print(f"RMSE : {rmse:.4f}")
print(f"MAE  : {mae:.4f}")

```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import KFold
import optuna 
scaler = StandardScaler()
X_train_scaled= scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

```

```python

kf = KFold(n_splits=3, shuffle=True, random_state=42)

def objective_Kfold_withScaled(trial):
    # Hyperparameters to tune
    params = {
        'iterations': trial.suggest_int('iterations', 1800, 3900, step=100),
        'learning_rate': trial.suggest_float('learning_rate', 0.02, 0.04, log=True),
        
        'depth': 4,
        
        'l2_leaf_reg': trial.suggest_int('l2_leaf_reg', 1, 10),
        
        
        'task_type': 'CPU',   
        'verbose': 0,
        'random_seed': 42
    }
    
    # Manual CV loop

    r2_scores = []
    for train_idx, valid_idx in kf.split(X_train_scaled):
        X_tr, X_val = X_train_scaled[train_idx], X_train_scaled[valid_idx]
        y_tr, y_val = y_train[train_idx], y_train[valid_idx]

        model = CatBoostRegressor(**params)
        model.fit(
            X_tr, y_tr,
            eval_set=(X_val, y_val),
            early_stopping_rounds=100,
            verbose=0
        )

        y_pred = model.predict(X_val)
        r2_scores.append(r2_score(y_val, y_pred))

    return np.mean(r2_scores)
```

```python
# 3️⃣ Run Optuna study
# --------------------------
study = optuna.create_study(direction="maximize", study_name = "Catboost_study_CPU", sampler=optuna.samplers.TPESampler(seed=42))
study.optimize(objective_Kfold_withScaled, n_trials=50, show_progress_bar=True, n_jobs=-1)
```

```python
print("Best mean CV R²:", study.best_value)
print("Best hyperparameters:", study.best_params)
```

```python
best_params = {'iterations': 3900, 'learning_rate': 0.03617035352979516, 'l2_leaf_reg': 1}
print(best_params)
```

```python
# 4️⃣ Train final model on all augmented train data with best params
# --------------------------
best_params = best_params

FinalCatBoostModel = CatBoostRegressor(
    **best_params,
    depth=4, 
    
    task_type= 'CPU',   # use 'CPU' for stable CV, switch to 'GPU' if desired
    verbose= 100,
    random_seed= 42,
)

FinalCatBoostModel.fit(X_train_scaled, y_train)

y_pred_log = FinalCatBoostModel.predict(X_test_scaled)

import numpy as np
from sklearn.metrics import r2_score

y_pred_real = np.expm1(y_pred_log)
y_test_real = np.expm1(y_test)

print("Final R² (log scale) | After hyperparameter tuning | CatBoost Regressor :", r2_score(y_test, y_pred_log))
print("Final R² (real scale) | After hyperparameter tuning | CatBoost Regressor:", r2_score(y_test_real, y_pred_real))


# Back-transform predictions and true values
y_pred_real = np.expm1(y_pred_log)
y_test_real = np.expm1(y_test)
print("R² for DOC on normal:", r2_score(y_test_real, y_pred_real))

# ---- Metrics on normal scale ----
mse  = mean_squared_error(y_test_real, y_pred_real)
rmse = np.sqrt(mse)
mae  = mean_absolute_error(y_test_real, y_pred_real)
r2   = r2_score(y_test_real, y_pred_real)

print("Final metrics on NORMAL scale (after inverse log):")
print(f"R²   : {r2:.4f}")
print(f"MSE  : {mse:.4f}")
print(f"RMSE : {rmse:.4f}")
print(f"MAE  : {mae:.4f}")
```

```python
kf = KFold(n_splits=3, shuffle=True, random_state=42)

def objective_Kfold_withScaled_NM(trial):
    # Hyperparameters to tune
    params = {
        'iterations': trial.suggest_int('iterations', 1800, 3900, step=100),
        'learning_rate': trial.suggest_float('learning_rate', 0.02, 0.04, log=True),
        
        'depth': 4,
        
        'l2_leaf_reg': trial.suggest_int('l2_leaf_reg', 1, 10),
        
        
        'task_type': 'CPU',   
        'verbose': 0,
        'random_seed': 42
    }
    
    # Manual CV loop
    r2_scores = []
    for train_idx, valid_idx in kf.split(X_train_scaled):
        X_tr, X_val = X_train_scaled[train_idx], X_train_scaled[valid_idx]
        y_tr, y_val = y_train[train_idx], y_train[valid_idx]

        model = CatBoostRegressor(**params)
        model.fit(
            X_tr, y_tr,
            eval_set=(X_val, y_val),
            early_stopping_rounds=100,
            verbose=0
        )

        y_pred = model.predict(X_val)
        r2_scores.append(r2_score(y_val, y_pred))

    return np.mean(r2_scores)
```

```python
study = optuna.create_study(direction="maximize", study_name = "Catboost_study_CPU", sampler=optuna.samplers.TPESampler(seed=42))
study.optimize(objective_Kfold_withScaled_NM, n_trials=50, show_progress_bar=True, n_jobs=-1)
```
